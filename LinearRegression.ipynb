# California Housing Price Prediction

A complete machine learning workflow to predict median house values in California using a **Linear Regression** model. This serves as a clean baseline benchmark for regression tasks.

## Workflow
1. **Import libraries** — pandas, numpy, matplotlib, seaborn, scikit-learn
2. **Load & explore dataset** — California Housing from `sklearn.datasets`
3. **Data preparation** — split into train / test sets (67% / 33%)
4. **Model training** — fit a `LinearRegression` model
5. **Evaluation & diagnostics** — scatter plot, residual KDE, R² score, RMSE

---

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

sns.set_style('whitegrid')
print('Libraries loaded successfully.')
```

## 1. Load & Explore the Dataset
We load the California Housing dataset from `scikit-learn` and inspect its structure.

```python
housing = fetch_california_housing()

hos = pd.DataFrame(housing.data, columns=housing.feature_names)
hos['PRICE'] = housing.target

print('Dataset shape:', hos.shape)
print('\nFeatures:', list(housing.feature_names))
print('\nFirst 5 rows:')
hos.head()
```

```python
# Basic statistics
hos.describe()
```

```python
# Check for missing values
print('Missing values per column:')
print(hos.isnull().sum())
```

## 2. Data Preparation
We separate the independent features ($X$) from the target variable ($Y$, the house price) and split them into training and testing sets (67% train / 33% test).

```python
X = hos.drop('PRICE', axis=1)
Y = hos['PRICE']

X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.33, random_state=5
)

print('Training set size :', X_train.shape)
print('Test set size     :', X_test.shape)
```

## 3. Model Training & Predictions
We fit a `LinearRegression` model on the training data and generate predictions on the held-out test set.

```python
lm = LinearRegression()
lm.fit(X_train, Y_train)
Y_pred = lm.predict(X_test)

print('Model trained successfully.')
print('Intercept :', round(lm.intercept_, 4))
```

## 4. Evaluation & Diagnostics
### 4a. Actual vs Predicted Prices
A scatter plot of actual versus predicted values. Points close to the diagonal line indicate good predictions.

```python
plt.figure(figsize=(7, 5))
plt.scatter(Y_test, Y_pred, alpha=0.4, edgecolors='k', linewidths=0.3)

# Perfect-prediction reference line
lims = [min(Y_test.min(), Y_pred.min()), max(Y_test.max(), Y_pred.max())]
plt.plot(lims, lims, 'r--', label='Perfect prediction')

plt.xlabel(r'Actual prices: $Y_i$')
plt.ylabel(r'Predicted prices: $\hat{Y}_i$')
plt.title(r'Actual vs Predicted Prices')
plt.legend()
plt.tight_layout()
plt.show()
```

### 4b. Residual Distribution
A KDE plot of the residuals ($Y_i - \hat{Y}_i$). A distribution centered around zero with small spread indicates a well-calibrated model.

```python
residuals = Y_test - Y_pred

plt.figure(figsize=(7, 4))
sns.kdeplot(np.array(residuals), bw_method=0.5, fill=True)
plt.axvline(0, color='r', linestyle='--', label='Zero residual')
plt.xlabel('Residual')
plt.title('Residuals Distribution')
plt.legend()
plt.tight_layout()
plt.show()
```

### 4c. Performance Metrics
Quantitative summary of model quality.

```python
r2   = r2_score(Y_test, Y_pred)
rmse = np.sqrt(mean_squared_error(Y_test, Y_pred))
mae  = np.mean(np.abs(residuals))

print(f'R² Score : {r2:.4f}')
print(f'RMSE     : {rmse:.4f}')
print(f'MAE      : {mae:.4f}')
```
